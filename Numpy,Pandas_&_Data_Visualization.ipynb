```python
import numpy as np

marks = np.array([85, 72, 90, 65, 88, 76, 95, 81, 69, 92])

print("Marks:", marks)
print("Total Marks:", np.sum(marks))
print("Average Marks:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))

print("Marks greater than 75:", marks[marks > 75])
```

NumPy Temperature Analysis

```python
import numpy as np

temperature = np.array([28, 31, 29, 32, 30, 33, 27])

print("Temperatures:", temperature)
print("Average Temperature:", np.mean(temperature))
print("Highest Temperature:", np.max(temperature))
print("Lowest Temperature:", np.min(temperature))

print("Days with temperature above 30°C:", np.where(temperature > 30)[0] + 1)

updated_temperature = temperature + 2

print("Updated Temperatures:", updated_temperature)
```

Pandas Student Performance


```python
import pandas as pd

data = {
    "Name": ["Arun", "Priya", "Rahul", "Sneha", "Karthik", "Divya", "Vijay", "Anu"],
    "Department": ["CSE", "ECE", "CSE", "IT", "CSE", "ECE", "IT", "CSE"],
    "Marks": [85, 92, 68, 78, 88, 72, 95, 65],
    "Attendance": [90, 85, 75, 82, 95, 78, 88, 70]
}

df = pd.DataFrame(data)

print("First 5 Students:")
print(df.head())

print("\nAverage Marks:")
print(df["Marks"].mean())

print("\nStudents who scored more than 75:")
print(df[df["Marks"] > 75])

print("\nStudents whose attendance is below 80%:")
print(df[df["Attendance"] < 80])

print("\nStudents sorted by Marks:")
print(df.sort_values("Marks"))
```

Pandas Product Sales Analysis

```python
import pandas as pd

data = {
    "Product Name": ["Laptop", "Mobile", "Headphones", "Keyboard", "Monitor"],
    "Category": ["Electronics", "Electronics", "Accessories", "Accessories", "Electronics"],
    "Price": [50000, 20000, 2000, 1500, 12000],
    "Quantity Sold": [20, 60, 80, 40, 55]
}

df = pd.DataFrame(data)

df["Total Sales"] = df["Price"] * df["Quantity Sold"]

print("Product Sales Data:")
print(df)

print("\nProduct with Highest Sales:")
print(df.loc[df["Total Sales"].idxmax()])

print("\nAverage Product Price:")
print(df["Price"].mean())

print("\nProducts with Quantity Sold greater than 50:")
print(df[df["Quantity Sold"] > 50])

print("\nProducts sorted by Total Sales:")
print(df.sort_values("Total Sales", ascending=False))
```

Monthly Sales Visualization

```python
import matplotlib.pyplot as plt

months = ["January", "February", "March", "April", "May", "June"]
sales = [25000, 30000, 28000, 35000, 40000, 45000]

plt.figure(figsize=(8,5))
plt.plot(months, sales, marker='o', label="Monthly Sales")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.show()

plt.figure(figsize=(8,5))
plt.bar(months, sales)
plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

Student Marks Visualization

```python
import matplotlib.pyplot as plt

students = ["Arun", "Priya", "Rahul", "Sneha", "Karthik",
            "Divya", "Vijay", "Anu", "Ravi", "Meena"]

marks = [85, 92, 68, 78, 88, 72, 95, 65, 55, 38]

plt.figure(figsize=(10,5))
plt.bar(students, marks)
plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.xticks(rotation=45)
plt.show()

excellent = sum(80 <= mark <= 100 for mark in marks)
good = sum(60 <= mark <= 79 for mark in marks)
average = sum(40 <= mark <= 59 for mark in marks)
needs_improvement = sum(mark < 40 for mark in marks)

categories = ["Excellent", "Good", "Average", "Needs Improvement"]
counts = [excellent, good, average, needs_improvement]

plt.figure(figsize=(7,7))
plt.pie(counts, labels=categories, autopct="%1.1f%%")
plt.title("Student Performance Distribution")
plt.show()
```
