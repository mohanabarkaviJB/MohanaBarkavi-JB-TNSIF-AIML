```python
# Student Marks Analyzer

students = {
    101: ("Arun", "CSE", 85),
    102: ("Priya", "ECE", 92),
    103: ("Rahul", "CSE", 78),
    104: ("Sneha", "IT", 92),
    105: ("Karthik", "CSE", 88)
}

# List of marks
marks = [85, 92, 78, 92, 88]

# Traversal
print("Student Marks:")
for mark in marks:
    print(mark)

# Linear Search
target = int(input("\nEnter a mark to search: "))

found = False

for mark in marks:
    if mark == target:
        found = True
        break

if found:
    print("Mark found!")
else:
    print("Mark not found!")

# Set - unique marks
unique_marks = set(marks)

print("\nUnique Marks:")
print(unique_marks)

# Dictionary - student ID to student details
student_id = int(input("\nEnter Student ID: "))

if student_id in students:
    print("Student Details:", students[student_id])
else:
    print("Student not found!")

# Tuple
student = students[101]
print("\nTuple Example:")
print(student)

# Array-like indexing
print("\nFirst mark:", marks[0])
print("Third mark:", marks[2])
```
